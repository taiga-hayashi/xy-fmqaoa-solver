# Paper Figures Notebook (`mlfilter-paper-repro`)

`analysis/mlfilter-paper-repro/<smoke|full>/` の集計 CSV を確認し、全 8 枚のプロット（`Boltzmann` 比較除外済み）を再生成・プレビューします。

```python
import os
import sys
from pathlib import Path
from IPython.display import Image, display

STUDY_DIR = Path.cwd()
if not (STUDY_DIR / 'plot_paper.py').exists():
    STUDY_DIR = Path('experiments/mlfilter-paper-repro').resolve()
sys.path.insert(0, str(STUDY_DIR))

from repro import default_output_dirs
import plot_paper as P

# full が存在すれば full、無ければ smoke を既定で参照（環境変数 MLFILTER_FIG_DIR で上書き可）
default_full = default_output_dirs('full')[1]
default_smoke = default_output_dirs('smoke')[1]
FIG_DIR = Path(os.environ.get('MLFILTER_FIG_DIR', default_full if (default_full / 'csv').exists() else default_smoke))
print('Target figure directory:', FIG_DIR)
```

```python
drawn = P.draw_all_figures(FIG_DIR)
print('Rendered figures:', drawn)
for stem in drawn:
    png_path = FIG_DIR / 'png' / f'{stem}.png'
    if png_path.exists():
        print(f'=== {stem} ===')
        display(Image(filename=str(png_path), width=620))
```
